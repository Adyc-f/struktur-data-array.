membuat array

```python
# Program untuk pengelolaan nilai mahasiswa
# dibuat untuk memenuhi tugas struktur data

nilai = []  # array kosong untuk menampung nilai nantinya

print("=== Input Nilai Mahasiswa ===")
```

Input 10 nilai array

```python
for i in range(10):
    n = int(input(f"Masukkan nilai ke-{i+1}: "))
    nilai.append(n)
```

```python
print("\nData nilai:", nilai)
```

Mencari nilai tertinggi dan terendah

```python
# mencari nilai tertinggi
maks = nilai[0]  # saya gunakan nilai pertama sebagai acuan awal

for n in nilai:
    if n > maks:
        maks = n

# mencari nilai terendah
minim = nilai[0]  # acuan awal

for n in nilai:
    if n < minim:
        minim = n

print("\nNilai tertinggi:", maks)
print("Nilai terendah:", minim)
```

Menghitung rata rata

```python
# menghitung total nilai
total = 0

for n in nilai:
    total += n  # menjumlahkan satu per satu

# menghitung rata-rata
rata_rata = total / len(nilai)

print("\nRata-rata nilai:", rata_rata)
```

Menghitung jumlah mahasiswa lulus

```python
# menghitung jumlah mahasiswa lulus
lulus = 0

for n in nilai:
    if n >= 60:
        lulus += 1

print("Jumlah mahasiswa lulus:", lulus)
```

Grafik nilai tertinggi dan terendah

```python
import matplotlib.pyplot as plt

# data untuk grafik
kategori = ['Tertinggi', 'Terendah']
nilai_grafik = [maks, minim]

plt.figure()
plt.bar(kategori, nilai_grafik)

plt.title("Perbandingan Nilai Tertinggi dan Terendah")
plt.ylabel("Nilai")

plt.show()
```

Grafik kelulusan

```python
# menghitung tidak lulus
tidak_lulus = len(nilai) - lulus

kategori2 = ['Lulus', 'Tidak Lulus']
jumlah = [lulus, tidak_lulus]

plt.figure()
plt.bar(kategori2, jumlah)

plt.title("Data Kelulusan Mahasiswa")
plt.ylabel("Jumlah Mahasiswa")

plt.show()
```

Hasil gabungan code

```python
print("\n=== HASIL ===")

print("Data nilai        :", nilai)
print("Nilai tertinggi   :", maks)
print("Nilai terendah    :", minim)
print("Rata-rata         :", rata_rata)
print("Jumlah lulus      :", lulus)

tidak_lulus = len(nilai) - lulus
print("Jumlah tidak lulus:", tidak_lulus)
```
